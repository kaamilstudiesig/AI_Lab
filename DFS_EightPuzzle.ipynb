```python
import copy
from collections import deque

start = [[2, 8, 3],
         [1, 6, 4],
         [7, 0, 5]]

goal = [[1, 2, 3],
        [8, 0, 4],
        [7, 6, 5]]

def solve_bfs(start_board):
    queue = deque([(start_board, 0)])
    visited = {str(start_board)}

    while queue:
        board, level = queue.popleft()

        if board == goal:
            print("Goal found")
            print(f"Level: {level}")
            print("Final Board:")
            for row in board:
                print(row)
            return True

        x, y = 0, 0
        for i in range(3):
            for j in range(3):
                if board[i][j] == 0:
                    x, y = i, j
                    break
            else:
                continue
            break

        for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            nx, ny = x + dx, y + dy

            if 0 <= nx < 3 and 0 <= ny < 3:
                next_board = [row[:] for row in board]

                next_board[x][y], next_board[nx][ny] = next_board[nx][ny], next_board[x][y]

                key = str(next_board)
                if key not in visited:
                    visited.add(key)
                    queue.append((next_board, level + 1))

    print("No solution")
    return False

solve_bfs(start)
```
